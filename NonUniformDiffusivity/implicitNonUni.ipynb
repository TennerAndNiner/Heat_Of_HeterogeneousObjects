```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
time_step = 0.25
step_size = 0.25
length = 1
```

```python
r = (time_step) / (step_size)**2
```

```python
x_range = np.arange(0, length + step_size, step_size)
t_range = np.arange(0, length + step_size, time_step)
```

```python
x_range
```

```python
x_len, t_len = len(x_range), len(t_range)
```

```python
x_len
```

There must be diffusivity values between 0 to the length of the rod.

```python
diffs = [1, 0.5, 1, 0.5, 1, 0.5]
```

```python
diffs
```

```python
def diffFunc(arr, i, r):
    return 1 + r * (arr[i - 1] + arr[i])
```

The main diagonal with 1 + r $D_{i-1}$ + r $D_i$

```python
main_diag = []

for i in range(1, x_len + 1):
    print(i)
    main_diag.append(diffFunc(arr = diffs, i = i, r = r))
```

```python
main_diag
```

The top diagonal is $-rD_{i}$ for $1 \leq i \leq m$

```python
top_diag = []
for j in range(0, x_len - 1):
    top_diag.append(-r * diffs[j])
```

```python
top_diag
```

Establish matrix A

```python
A = np.zeros((x_len, t_len))
A
```

```python
A = np.diag(main_diag, k = 0) + np.diag(top_diag, k = -1) + np.diag(top_diag, k = 1)
A
```

```python
n = 1
```

```python
grid = np.zeros((x_len, t_len))
grid[:,0] = 10
#grid[0, :] = 0
#grid[-1, :] = 0
grid
```

```python
for i in range(0, t_len - 1):
    b = grid[:, i]
    x = np.linalg.solve(A, b)
    # Set x as the new 
    grid[:, i + 1] = x
```
