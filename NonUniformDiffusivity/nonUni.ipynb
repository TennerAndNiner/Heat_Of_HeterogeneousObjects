```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
time_step = 0.05
step_size = 0.10
length = 1
```

L-Stable, A-Stable schemes

```python
r = (time_step) / (step_size)**2
```

```python
x_range = np.arange(0, length + step_size, step_size)
t_range = np.arange(0, length + time_step, time_step)
```

```python
x_len, t_len = len(x_range), len(t_range)
```

```python
def diffFunc(i):
     if i < (x_len / 2):
         return 0.1
     else:
         return 0.5
```

```python
grid = np.zeros((x_len, t_len))
grid
```

```python
n = 1
```

```python
grid[:,0] = np.sin(n * np.pi * x_range)
#grid[0, :] = 0
#grid[-1, :] = 0
#grid
```

```python
for j in range(0, t_len - 1):
    for i in range(1, x_len - 1):
        grid[i, j + 1] = (r * diffFunc(i) * grid[i + 1, j]) + (r * diffFunc(i - 1) * grid[i - 1, j]) + ((1 - r * (diffFunc(i) + diffFunc(i-1))) * grid[i, j])
        print(i, j)
grid.round(7)
```
